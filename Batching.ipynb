### **Batching**

```python
%load_ext dotenv
%dotenv
```

```python
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate
```

```python
chat_template = ChatPromptTemplate.from_messages([('human', 
                                                    "I've recently adopted a {pet} which is a {breed}. Could you suggest several training tips?")])
```

```python
chat = ChatOpenAI(
    model = 'gpt-4o',
    seed = 7,
    temperature = 0,
    max_completion_tokens = 100
)
```

```python
chain = chat_template | chat
```

```python
chain.invoke({'pet':'dog', 'breed':'shepherd'})
```

```python
chain.batch([{'pet':'dog', 'breed':'shepherd'}, {'pet':'dragon', 'breed':'night fury'}])
```

```python
%%time
chain.batch([{'pet':'dog', 'breed':'shepherd'}, {'pet':'dragon', 'breed':'night fury'}])
```

```python
%%time
chain.invoke({'pet':'dog', 'breed':'shepherd'})
```

```python
%%time
chain.invoke({'pet':'dragon', 'breed':'night fury'})
```

Using a batch() method is more efficient than a sequence of invoke() methods
