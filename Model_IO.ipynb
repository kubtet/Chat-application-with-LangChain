### **ChatOpenAI**

```python
%load_ext dotenv
%dotenv
```

```python
from langchain_openai.chat_models import ChatOpenAI
```

```python
chat = ChatOpenAI(model_name = 'gpt-4o', 
                  seed = 365,
                  temperature = 0
                 )
```

```python
''' I've recently adopted a dog. 
Could you suggest some dog names? '''

# Triple ' allows to break the lines
```

```python
response = chat.invoke(''' I've recently adopted a dog. Could you suggest some dog names? ''')
```

```python
print(response.content)
```

```python
response
```

```python
chat = ChatOpenAI(model_name = 'gpt-4o', 
                  seed = 365,
                  temperature = 0,
                  max_completion_tokens = 100
                 )

response = chat.invoke(''' I've recently adopted a dog. Could you suggest some dog names? ''')
```

```python
print(response.content)
```

### **System and Human Messages**

```python
from langchain_core.messages import SystemMessage, HumanMessage
```

```python
message_h = HumanMessage(content = ''' I've recently adopted a cat. Can you suggest some cat names? ''')
```

```python
response = chat.invoke([message_h])
```

```python
response
```

```python
print(response.content)
```

```python
message_s = SystemMessage(content = ''' You are Marv, a chatbot that reluctantly answers questions with sarcastic responses. ''')
```

```python
response = chat.invoke([message_s, message_h])
```

```python
print(response.content)
```

### **AI Messages**

```python
from langchain_core.messages import AIMessage
```

##### Ideally is to have a System Message and couple of pairs Human - AI Messages

```python
message_h_dog = HumanMessage(content = ''' I've recently adopted a dog. Can you suggest some dog names? ''')
message_ai_dog = AIMessage(content = ''' Oh sure, because naming a dog is such a difficult task. How about "Dog"? 
Or maybe "Canine"? If you're feeling adventurous, you could even go with "Fido" or "Spot." But hey, don't let me stifle your creativity. ''')

message_h_cat = HumanMessage(content = ''' I've recently adopted a cat. Can you suggest some cat names? ''')
message_ai_cat = AIMessage(content = ''' Oh, sure, because naming a cat is such a monumental task. How about "Whiskers" or "Fluffy"? 
Those are totally original. Or maybe go wild and name it "Chair" or "Lamp." I'm sure your cat will appreciate the creativity. ''')

message_h_fish = HumanMessage(content = ''' I've recently adopted a fish. Can you suggest some fish names? ''')
```

```python
response = chat.invoke([message_h_dog, message_ai_dog, message_h_cat, message_ai_cat, message_h_fish])
```

```python
print(response.content)
```
