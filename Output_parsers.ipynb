### **String Output Parser**

```python
%load_ext dotenv
%dotenv
```

```python
from langchain_openai.chat_models import ChatOpenAI
from langchain_core.messages import HumanMessage
from langchain_core.output_parsers import StrOutputParser
```

```python
chat = ChatOpenAI(model_name = 'gpt-4o',
                 seed = 7,
                 temperature = 0,
                 max_completion_tokens = 100)
```

```python
message_h = HumanMessage(content = "Can you give me an interesting fact that I probably didn't know about?")
```

```python
response = chat.invoke([message_h])
```

```python
response
```

```python
str_output_parser = StrOutputParser()
```

```python
response_parsed = str_output_parser.invoke(response)
response_parsed
```

### **Comma-Separated List Output Parser**

```python
message_h = HumanMessage(content = f''' I've recently adopted a dog. Could you suggest me some dog names? 

{CommaSeparatedListOutputParser().get_format_instructions()}
''')
```

```python
from langchain_core.output_parsers import CommaSeparatedListOutputParser
```

```python
response = chat.invoke([message_h])
print(response.content)
```

```python
list_output_parser = CommaSeparatedListOutputParser()
```

```python
response_parsed = list_output_parser.invoke(response)
```

```python
response_parsed
```

```python
print(list_output_parser.get_format_instructions())
```
