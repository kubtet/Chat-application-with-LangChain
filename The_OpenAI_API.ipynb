```python
%load_ext dotenv
%dotenv
```

```python
import os
import openai
```

```python
openai.api_key = os.getenv('OPENAI_API_KEY')
```

```python
client = openai.OpenAI()
```

```python
completion = client.chat.completions.create(model = 'gpt-5', 
                                            messages = [{'role': 'system', 
                                                       'content': ''' You are Marv, a chatbot that reluctantly
                                                       answers questions with sarcastic responses. '''},
                                                      {'role': 'user',
                                                        'content': ''' I've recently adopted a dog. Could you suggest some dog names? '''}])
```

```python
completion
```

#### Parameter **n** -> number of responses = number of tokens

```python
print(completion.choices[0].message.content)
```

### Tweet classifier which classifies comments as positive, negative or neutral

```python
tweet_classifier = client.chat.completions.create(model='gpt-5',
                                                messages=[{'role': 'system', 'content': ''' You are a tweet classifier that classifies comments as positive, negative, or neutral. '''},
                                                         {'role': 'user', 'content': ''' I have never seen prettier dress! '''}])
```

```python
print(tweet_classifier.choices[0].message.content)
```

### Temperature, Max Tokens, and Streaming

```python
completion = client.chat.completions.create(model = 'gpt-4o', 
                                            messages = [{'role': 'system', 
                                                       'content': ''' You are Marv, a chatbot that reluctantly
                                                       answers questions with sarcastic responses. '''},
                                                      {'role': 'user',
                                                        'content': ''' Could you explain briefly what a black hole is? '''}],
                                           max_completion_tokens = 250)
```

```python
print(completion.choices[0].message.content)
```

```python
completion = client.chat.completions.create(model = 'gpt-4o', 
                                            messages = [{'role': 'user',
                                                        'content': ''' Could you explain briefly what a black hole is? '''}],
                                           max_completion_tokens = 250,
                                           temperature = 0)
```

```python
print(completion.choices[0].message.content)
```

```python
completion = client.chat.completions.create(model = 'gpt-4o', 
                                            messages = [{'role': 'user',
                                                        'content': ''' Could you explain briefly what a black hole is? '''}],
                                           max_completion_tokens = 250,
                                           temperature = 0)
```

```python
print(completion.choices[0].message.content)
```

```python
# Both results created with the temperature parameter = 0 are very similar unless the same
```

```python
completion = client.chat.completions.create(model = 'gpt-4o', 
                                            messages = [{'role': 'user',
                                                        'content': ''' Could you explain briefly what a black hole is? '''}],
                                           max_completion_tokens = 250,
                                           temperature = 2)
```

```python
print(completion.choices[0].message.content)
```

```python
completion = client.chat.completions.create(model = 'gpt-4o', 
                                            messages = [{'role': 'user',
                                                        'content': ''' Could you explain briefly what a black hole is? '''}],
                                           max_completion_tokens = 250,
                                           temperature = 0,
                                           seed = 365)
```

```python
completion
```

```python
completion = client.chat.completions.create(model = 'gpt-4o', 
                                            messages = [{'role': 'user',
                                                        'content': ''' Could you explain briefly what a black hole is? '''}],
                                           max_completion_tokens = 250,
                                           temperature = 0,
                                           seed = 365,
                                           stream = True)
```

```python
completion
```

```python
for i in completion:
    print(i.choices[0].delta.content, end="")
```
